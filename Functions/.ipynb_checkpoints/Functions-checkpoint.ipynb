```python
def reverse(s):
    result = ""
    for c in s:
        result = c + result
    return result
t = input("Please enter a string: ")
print("The reverse of", t, "is", reverse(t) )
```

#### Practice 5.1 Write a function called explode that gives a string a list of the characters of the string.

```python
def explode():
    results = []
    t = input("Please enter a string: ")
    for line in t:
        words = line.split()
        results.append(words)
    return results
results = explode()
print(results)
```

```python
import turtle
def drawfigure(turtle, *args):
    for i in range(0, len(args),2):
        turtle.forwad(args[i])
        turtle.left(args[i+1])
def main():
    t = turtle.Turtle()
    screen = t.getscreen()
    drawFigure(t, 50,90,30,50,90,30,90)
    screen.exitonclick()
if __name__ == " __main__ ":
    main()
```

```python
def IsEven():
    while True:
        try:
            num = float(input("Enter a value"))
            break
        except ValueError:
            print("Enter a valid Numeric Value")
    if num % 2 == 0:
        print("True")
    else:
        print("False")
IsEven()

        
    
```

 - Write a function called allEvens that gives a list of integers, returns a new list containing only the even integers. Use the function in a program and test your code on several different values.

```python
def allEvens():
    lst = []
    for i in range(5):
        while True:
            try:
                num = float(input("Enter a value"))
                lst.append(num)
                break
            except ValueError:
                print("Enter a valid Numerical value ")
    final = []
    for line in lst:
        if line % 2 == 0:
            final.append(line)
    print(final)
allEvens()
    
                
    
                
            
    
```
