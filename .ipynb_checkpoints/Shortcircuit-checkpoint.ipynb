```python
lst = []
# Loop exactly 5 times
for i in range(5):
    while True:
        try:
            num = float(input("Enter a number: "))
            lst.append(num)
            break
        except ValueError:
            print("Enter a Valid Number")
# Check if the list is empty before finding min/max
smallest = None
biggest = None
for number in lst:
    if smallest is None or number < smallest:
        smallest = number
    if biggest is None or number > biggest:
        biggest = number
print(smallest)
print(biggest)
```
